```python
# Uniform Distribution
import numpy as np                     # import numpy for random number generation
import matplotlib.pyplot as plt        # import matplotlib for plotting

# values from 0 to 10
values = np.random.uniform(0, 10, 100000)   # generate 100000 uniform random values between 0 and 10

# plotting a histogram – becomes more uniform with more values
plt.hist(values, bins=100, density=True, alpha=0.3)   # draw normalized histogram with 100 bins

plt.title("Continuous Uniform Distribution between 0 and 10")   # set plot title
plt.xlabel("x")                                   # set x-axis label
plt.ylabel("Probability Density")                 # set y-axis label
plt.grid(True)                                    # enable grid

plt.show()                                        # display the plot
```

```python
# Normal Distribution
import numpy as np                          # import numpy
import matplotlib.pyplot as plt             # import matplotlib
from scipy.stats import norm                # import normal distribution

# Parameters
mu = 70                                     # mean
sigma = 10                                  # standard deviation

# X-axis values
x = np.linspace(30, 110, 1000)              # create range of x values

# Normal PDF
y = norm.pdf(x, mu, sigma)                  # compute normal pdf values

# Plot the bell curve
plt.plot(x, y, color='black', linewidth=2, label='Normal Distribution')  # draw curve

# Shade 68% region (μ ± 1σ)
x1 = np.linspace(mu - sigma, mu + sigma, 1000)   # range for 1 sigma
plt.fill_between(x1, norm.pdf(x1, mu, sigma), color='green', alpha=0.3, label='68% region (1σ)')

# Shade 95% region (μ ± 2σ)
x2 = np.linspace(mu - 2*sigma, mu + 2*sigma, 1000)  # range for 2 sigma
plt.fill_between(x2, norm.pdf(x2, mu, sigma), color='yellow', alpha=0.2, label='95% region (2σ)')

# Shade 99.7% region (μ ± 3σ)
x3 = np.linspace(mu - 3*sigma, mu + 3*sigma, 1000)  # range for 3 sigma
plt.fill_between(x3, norm.pdf(x3, mu, sigma), color='red', alpha=0.1, label='99.7% region (3σ)')

# Labels and title
plt.title("Normal Distribution of Student Scores")  # plot title
plt.xlabel("Score")                                 # x label
plt.ylabel("Probability Density")                   # y label
plt.legend()                                        # show legend

plt.grid(True)                                      # show grid

plt.show()                                          # display plot
```

```python
# Plot y = √x
import numpy as np
import matplotlib.pyplot as plt

# Define x values (x >= 0 for sqrt)
x = np.linspace(0, 100, 100_000)
y = np.sqrt(x)

# Plot
plt.plot(x, y, label="y = √x")
plt.xlabel("x")
plt.ylabel("y")
plt.title("y = √x")
plt.legend()
plt.grid(True)

plt.show()
```

```python
# Create x values from -2π to 2π
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-2 * np.pi, 2 * np.pi, 100)
y = np.sin(x)

# Plot the graph
plt.plot(x, y, label="y = sin(x)", color="blue")

# Highlight x = 0 and y = 0 axes
plt.axvline(0, color="black", linewidth=1.5, linestyle="--", label="x = 0")
plt.axhline(0, color="black", linewidth=1.5, linestyle="--", label="y = 0")

# Labels and title
plt.xlabel("x")
plt.ylabel("y")
plt.title("Graph of y = sin(x)")
plt.grid(True)
plt.legend()

plt.show()
```
